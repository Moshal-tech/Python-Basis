Asks the user to enter two numbers.

Adds the two numbers.

Multiply two numbers

Divide two numbers

Floor divide two numbers

Find the remainder.

Prints the result with a message.

```python
print("This is a simple calculator.")
number1 = float(input("Enter number one:"))
number2 = float(input("Enter number two:"))

print(f"Addition {number1} + {number2} is {number1 + number2}.")

print(f"Multiplication {number1} * {number2} is {number1 * number2}.")

print(f"Division {number1} / {number2} is {number1 / number2}.")

print(f"Floor Division {number1} // {number2} is {number1 // number2}.")

print(f"Remainder {number1} % {number2} is {number1 % number2}.")
```
